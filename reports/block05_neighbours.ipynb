```python
d = pd.read_csv("data/processed/site_daily.csv", parse_dates=["date"])

# 16-day means: the scale of the satellite
d16 = d.set_index("date").groupby("site_id")[["GPP_NT_VUT_REF", "evi"]].resample("16D").mean().reset_index()

print(d16.groupby("site_id").apply(lambda g: g["GPP_NT_VUT_REF"].corr(g["evi"])))
```
